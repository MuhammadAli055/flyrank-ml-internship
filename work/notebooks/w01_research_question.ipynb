## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
## The Question: Decision, Action, and Cost of a Wrong Call

'''

**My research question:**
Which content pages should be prioritized for review, refresh, or improvement —
and in what order?

**Unit of analysis:**
One content page (one row = one page).

**Output:**
A ranked list of pages, ordered from most urgent to least urgent, with a score and
a reason code explaining why each page was flagged.

**The decision this improves:**
A content manager deciding which pages to spend their limited review time on this week.

**The action someone takes from it:**
They open the top pages on the ranked list and decide: should this page be rewritten,
updated, expanded, or just monitored? The list tells them WHERE to look — they still
use human judgment to decide WHAT to do.

**The cost of a wrong recommendation:**
- If the system ranks a page too HIGH (false alarm): a reviewer wastes time on a page
  that was actually fine. This wastes limited team capacity.
- If the system ranks a page too LOW (missed case): a page that was declining or losing
  clicks gets ignored, and the website loses traffic and visibility that could have been
  saved with a timely update.

**Why data and ML can help:**
A human cannot manually review thousands of pages every week. A scoring model reads
signals like impressions, clicks, position, content age, and engagement rate across
all pages at once — and surfaces the ones most likely to need attention. The starter
pipeline already showed this works better than a fixed hand-written rule.
'''
```

**What I cannot claim:**
